```python
# Завдання 4.1

import os
from sqlalchemy import create_engine, text
import pandas as pd
from dotenv import load_dotenv


load_dotenv()
engine = create_engine(os.getenv("SHOPDB_URL"))

top_orders = pd.read_sql("""
SELECT
	order_id, 
	created_at, 
	status, 
	total_amount
FROM shop.orders 
WHERE country_code = 'UA'
ORDER BY total_amount DESC, order_id
LIMIT 10;
""", engine)
print(top_orders)
```

```python
# Завдання 4.2

status_counts = pd.read_sql("""
SELECT status, COUNT(order_id) AS orders_cnt 
FROM shop.orders 
GROUP BY status
ORDER BY orders_cnt DESC, status;
""", engine)

print(status_counts)
```

```python
# Завдання 5.1

customer_orders = pd.read_sql(text("""
SELECT 
	order_id, 
	created_at, 
	status, 
	total_amount
FROM shop.orders
WHERE customer_id = :customer_id
ORDER BY created_at ASC;
"""), engine,
params={"customer_id": 101})

print(customer_orders )
```

```python
# Завдання 5.2

def orders_by_country_for_day(day_start, day_end, country_code):
    return pd.read_sql(text("""
        SELECT 
            order_id, 
            customer_id, 
            created_at, 
            total_amount
        FROM shop.orders
        WHERE created_at >= :day_start
            AND created_at < :day_end
            AND country_code = :country_code
        ORDER BY created_at, order_id;
    """), engine,
    params={"day_start": day_start, "day_end": day_end, "country_code":country_code})

pl_orders = orders_by_country_for_day('2025-12-01', '2025-12-02', 'PL')
print(pl_orders)

ua_orders = orders_by_country_for_day('2025-12-01', '2025-12-02', 'UA')
print(ua_orders)
```
